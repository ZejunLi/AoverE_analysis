```python
energy_field = "trapEftp_cal"
aoe_field = "A_max_mw_o_E_Corrected"
```

```python
from pygama.pargen.survival_fractions import get_survival_fraction
```

```python
ak_cal = obj2_cal.view_as("ak")
energy = ak.to_numpy(ak_cal["trapEftp_cal"])
aoe = ak.to_numpy(ak_cal["A_max_mw_o_E_Corrected"])
```

```python
low_cut = 0.70   # 换成你的真实值

sf_dep, err_dep, values_dep, errors_dep = get_survival_fraction(
    energy=energy,
    cut_param=aoe,
    cut_val=low_cut,
    peak=1592.5,
    eres_pars=1.0,
    fit_range=(1540, 1610),
    mode="greater",
    display=0,   # noplot
)

print(f"DEP SF = {sf_dep:.2f} +/- {err_dep:.2f} %")
print("epsilon_sig =", values_dep["epsilon_sig"])
```
