```python
from scipy.signal import find_peaks

index = 0          # 0 = ch1, 1 = ch2
dataset_offset = 0 # 0=cal, 2=bg, 4=signal

spec = dataset_offset + index

k42_window = (centers[spec] > 1400) & (centers[spec] < 1550)
tl208_window = (centers[spec] > 2500) & (centers[spec] < 3000)

k42_slice = hist[spec][k42_window]
k42_peak_idx = np.argmax(k42_slice)
k42_peak_adc = centers[spec][k42_window][k42_peak_idx]
print("K42 peak ADC:", k42_peak_adc)

tl208_slice = hist[spec][tl208_window]
FEP_peak_idx = np.argmax(tl208_slice)
tl208_peak_adc = centers[spec][tl208_window][FEP_peak_idx]
print("Tl208 peak ADC:", tl208_peak_adc)

E_k42   = 1461.0
E_tl208 = 2614.5
E_dep   = 1592.5

adc_peaks = np.array([k42_peak_adc, tl208_peak_adc])
keV_peaks = np.array([E_k42, E_tl208])
```

```python
# predicted DEP window
# dep_window = (centers[spec] > dep_adc_pred-60) & (centers[spec] < dep_adc_pred+60)

dep_window = (centers[spec] > 1590-50) & (centers[spec] < 1590+50)

dep_slice = hist[spec][dep_window]
dep_peak_idx = np.argmax(dep_slice)

dep_peak_adc = centers[spec][dep_window][dep_peak_idx]

print("DEP peak ADC:", dep_peak_adc)

print(centers[spec][:10])

FEP_peak_adc = centers[spec][tl208_window][FEP_peak_idx]

print("FEP peak ADC:", FEP_peak_adc)
```

```python
from scipy.optimize import curve_fit
import numpy as np

# --- Fit function ---
def gauss_lin(x, A, mu, sigma, m, c):
    return A * np.exp(-(x - mu)**2 / (2 * sigma**2)) + m * x + c


def find_peaks_range(hist, centers, energy_window):

    hist_np = np.asarray(hist)
    centers_np = np.asarray(centers)

    # select region
    mask = energy_window
    x = centers_np[mask]
    y = hist_np[mask]

    # initial guesses
    p0 = [
        np.max(y),           # amplitude
        x[np.argmax(y)],     # peak position
        2.0,                 # sigma
        0.0,                 # slope
        np.min(y)            # offset
    ]

    popt, _ = curve_fit(gauss_lin, x, y, p0=p0)

    A, mu, sigma, m, c = popt

    # compute FWHM
    FWHM = 2.355 * sigma

    # peak range
    peak_min = mu - 1.5 * FWHM
    peak_max = mu + 1.5 * FWHM

    return peak_min, peak_max, popt, x, y
```

```python
def fit_and_plot_peak(hist, centers, window, title):

    peak_min, peak_max, popt, x, y = find_peaks_range(hist, centers, window)

    mu = popt[1]
    sigma = popt[2]
    FWHM = 2.355 * sigma

    peak_min = mu - 1.5 * FWHM
    peak_max = mu + 1.5 * FWHM

    plt.figure()
    plt.plot(x, y)
    plt.plot(x, gauss_lin(x, *popt))
    plt.axvline(peak_min)
    plt.axvline(peak_max)

    plt.xlabel("Energy (keV)")
    plt.ylabel("Counts")
    plt.title(title)

    plt.show()

    print(f"{title} peak at {mu:.3f} keV")
    print(f"sigma = {sigma:.3f} keV")
    print(f"FWHM = {FWHM:.3f} keV")
    print(f"Window: {peak_min:.3f} - {peak_max:.3f} keV")

    return peak_min, peak_max, popt
spec = dataset_offset + index

dep_min, dep_max, popt_dep = fit_and_plot_peak(
    hist[spec],
    centers[spec],
    dep_window,
    "DEP Gaussian Fit"
)

fep_min, fep_max, popt_fep = fit_and_plot_peak(
    hist[spec],
    centers[spec],
    tl208_window,
    "FEP Gaussian Fit"
)
```

```python
from scipy.special import erf
import numpy as np

def counts(popt, window_min, window_max):

    A, mu, sigma, m_fit, c_fit = popt

    # total Gaussian area
    N_total = A * sigma * np.sqrt(2*np.pi)

    # arguments for erf
    arg_max = (window_max - mu) / (np.sqrt(2) * sigma)
    arg_min = (window_min - mu) / (np.sqrt(2) * sigma)

    # Gaussian area inside window
    N_window = A * sigma * np.sqrt(np.pi/2) * (erf(arg_max) - erf(arg_min))

    fraction = N_window / N_total

    print("Gaussian counts in window:", N_window)
    print("Fraction kept:", fraction)

    return N_window, fraction
counts(popt_dep, dep_min, dep_max)
counts(popt_fep, fep_min, fep_max)
```

```python
DEP_mask = {}
tl208_mask = {}
Energy_mask = {}

E_upperlim = 2239
E_lowerlim = 1839

for dataset_offset in [0, 2, 4]:   # cal, bg, signal

    spec = dataset_offset + index

    DEP_mask[spec] = (
        (E_trapEftps[spec] > dep_min) &
        (E_trapEftps[spec] < dep_max)
    )

    tl208_mask[spec] = (
        (E_trapEftps[spec] > fep_min) &
        (E_trapEftps[spec] < fep_max)
    )

    Energy_mask[spec] = (
        (E_trapEftps[spec] > E_lowerlim) &
        (E_trapEftps[spec] < E_upperlim)
    )

print(E_upperlim, E_lowerlim)
```

```python
aoe_signal = {}
E_lowerlim = 1839
E_upperlim = 2239

for smoothing in smoothings:
    E_signal1 = ak_signal[smoothing+"_ch1"]["trapEftp_cal"]
    E_signal2 = ak_signal[smoothing+"_ch2"]["trapEftp_cal"]
    roi_mask1 = (E_signal1 > E_lowerlim) & (E_signal1 < E_upperlim)
    roi_mask2 = (E_signal2 > E_lowerlim) & (E_signal2 < E_upperlim)

    aoe_signal[smoothing+"_ch2"] = ak_signal[smoothing+"_ch2"]["Amax_o_E_Classifier"][roi_mask2]
    aoe_signal[smoothing+"_ch1"] = ak_signal[smoothing+"_ch1"]["Amax_o_E_Classifie"][roi_mask1]
```
