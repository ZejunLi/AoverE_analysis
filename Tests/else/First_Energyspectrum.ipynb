## TrapEftp without cuts

```python
# remove nans
E_ftp = ak.nan_to_none(ak_dsp["trapEftp"])
E_ftp = ak.drop_none(E_ftp)
E_ftp_np = ak.to_numpy(E_ftp)
# histogram
hist, edges = np.histogram(E_ftp_np, bins=3000)

# plot
centers = 0.5 * (edges[:-1] + edges[1:])
plt.figure(figsize=(7,5))
plt.step(centers, hist, where="mid")
plt.yscale("log")
plt.xlabel("trapEftp (ADC units)")
plt.ylabel("Counts")
plt.title("trapEftp spectrum (no cuts)")
plt.grid(alpha=0.3)
plt.show()
```

## compare max with ftp

```python
emin = min(E_max_np.min(), E_ftp_np.min())
emax = max(E_max_np.max(), E_ftp_np.max())

bins = 3000
hist_trap, edges = np.histogram(E_max_np, bins=bins, range=(emin, emax))
hist_ftp,  _     = np.histogram(E_ftp_np,  bins=bins, range=(emin, emax))

centers = 0.5 * (edges[:-1] + edges[1:])

plt.figure(figsize=(7,5))
plt.step(centers, hist_trap, where="mid", label="trapEmax", alpha=0.5)
plt.step(centers, hist_ftp,  where="mid", label="trapeftp", alpha=0.5)

plt.yscale("log")
plt.xlabel("Energy (ADC units)")
plt.ylabel("Counts")
plt.title("trapEmax vs trapeftp (no quality cuts)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
mask = (
    (ak_dsp["trapEftp"] > 0) &
    (ak_dsp["A_max"] > 0)
)

# 4. energy
E_ftp_cut = ak_dsp["trapEftp"][mask]
E_ftp_cut_np = ak.to_numpy(E_ftp_cut)
```

### before after cut comparison

```python
emin = min(E_ftp_np.min(), E_ftp_cut_np.min())
emax = max(E_ftp_np.max(), E_ftp_cut_np.max())

bins = 3000
hist_ftp, edges  = np.histogram(E_ftp_np, bins=bins, range=(emin, emax))
hist_ftp_cut,  _     = np.histogram(E_ftp_cut_np,  bins=bins, range=(emin, emax))

centers = 0.5 * (edges[:-1] + edges[1:])

plt.figure(figsize=(7,5))
plt.step(centers, hist_ftp, where="mid", label="trapeftp", alpha=0.5)
plt.step(centers, hist_ftp_cut,  where="mid", label="trapeftp (cut)", alpha=0.5)

plt.yscale("log")
plt.xlabel("Energy (ADC units)")
plt.ylabel("Counts")
plt.title("trapeftp before and after cuts")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
