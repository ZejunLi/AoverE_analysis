```python
from scipy.signal import find_peaks
index = 1 #0 for ch1 1for ch2
# Search windows (ADC units!)
k42_window = (centers[index] > 1400) & (centers[index] < 1550)
tl208_window = (centers[index] > 2500) & (centers[index] < 3000)

k42_slice = hist[index][k42_window]
k42_peak_idx = np.argmax(k42_slice)
k42_peak_adc = centers[index][k42_window][k42_peak_idx]
print("K42 peak ADC:", k42_peak_adc)
tl208_slice = hist[index][tl208_window]
FEP_peak_idx = np.argmax(tl208_slice)  # Find first peak in the slice
tl208_peak_adc = centers[index][tl208_window][FEP_peak_idx]
print("Tl208 peak ADC:", tl208_peak_adc)
E_k42   = 1524.7   # keV
E_tl208 = 2614.5   # keV

adc_peaks = np.array([k42_peak_adc, tl208_peak_adc])
keV_peaks = np.array([E_k42, E_tl208])

a, b = np.polyfit(adc_peaks, keV_peaks, 1)

print("Calibration:")
print("E_keV = {:.4f} * ADC + {:.2f}".format(a, b))
E_dep = 1592.5   # keV
dep_adc_pred = (E_dep - b) / a
def adc_to_keV(adc):
    return a * adc + b
def keV_to_adc(keV):
    return (keV - b) / a
print("Predicted DEP ADC:", dep_adc_pred)
```

```python
dep_window = (centers[index] > 1550) & (centers[index] < 1600)
dep_slice = hist[index][dep_window]
dep_peak_idx = np.argmax(dep_slice)
```

```python
from scipy.optimize import curve_fit

# --- Fit function ---
def gauss_lin(x, A, mu, sigma, m, c):
    return A*np.exp(-(x-mu)**2/(2*sigma**2)) + m*x + c
def find_peaks_range(hist,center,energy_range):
    # Convert to numpy
    hist_np = np.asarray(hist)
    centers_np = np.asarray(center)

    # Restrict region around DEP
    window = energy_range
    x = centers_np[window]
    y = hist_np[window]
    # Initial guesses
    p0 = [
        np.max(y),      # A
        x[np.argmax(y)],# mu
        2.0,            # sigma (~ few keV)
        0.0,            # slope
        np.min(y)       # offset
    ]

    popt, _ = curve_fit(gauss_lin, x, y, p0=p0)

    A, mu, sigma, m, c = popt
    # FWHM from sigma
    FWHM = 2.355 * sigma

    # Use wider window
    min = mu - 1.5 * FWHM
    max = mu + 1.5 * FWHM
    return min, max, popt,x,y
```

```python
dep_min,dep_max,popt,x,y = find_peaks_range(hist[index], centers[index], dep_window)
print(f"DEP peak at {popt[1]:.2f} keV")
print(f"Automatic window: {dep_min:.2f} - {dep_max:.2f} keV")
hist_np = np.asarray(hist[index])
centers_np = np.asarray(centers[index])
# --- Define window ---
FWHM = 2.355 * popt[2]  # sigma from fit
dep_min = popt[1] - 1.5 * FWHM
dep_max = popt[1] + 1.5 * FWHM

# --- Plot ---
plt.figure()
plt.plot(x, y)
plt.plot(x, gauss_lin(x, *popt))
plt.axvline(dep_min)
plt.axvline(dep_max)

plt.xlabel("Energy (keV)")
plt.ylabel("Counts")
plt.title("DEP Gaussian Fit with Selection Window")

plt.show()

print(f"mu = {popt[1]:.3f} keV")
print(f"sigma = {popt[2]:.3f} keV")
print(f"FWHM = {FWHM:.3f} keV")
print(f"Window: {dep_min:.3f} - {dep_max:.3f} keV")




fep_min,fep_max,popt2,x2,y2 = find_peaks_range(hist[index], centers[index], tl208_window)
print(f"FEP peak at {popt2[1]:.2f} keV")
print(f"Automatic window: {fep_min:.2f} - {fep_max:.2f} keV")
hist_np = np.asarray(hist[index])
centers_np = np.asarray(centers[index])
# --- Define window ---
FWHM = 2.355 * popt2[2]  # sigma from fit
fep_min = popt2[1] - 1.5 * FWHM
fep_max = popt2[1] + 1.5 * FWHM

# --- Plot ---
plt.figure()
plt.plot(x2, y2)
plt.plot(x2, gauss_lin(x2, *popt2))
plt.axvline(fep_min)
plt.axvline(fep_max)

plt.xlabel("Energy (keV)")
plt.ylabel("Counts")
plt.title("FEP Gaussian Fit with Selection Window")

plt.show()

print(f"mu = {popt2[1]:.3f} keV")
print(f"sigma = {popt2[2]:.3f} keV")
print(f"FWHM = {FWHM:.3f} keV")
print(f"Window: {fep_min:.3f} - {fep_max:.3f} keV")
```

```python
from scipy.special import erf
import numpy as np
def counts(popt,min,max):
    A, mu, sigma, m_fit, c_fit = popt

    # Gaussian total area
    N_total = A * sigma * np.sqrt(2*np.pi)

    # Gaussian area in window
    arg_max = (max - mu) / (np.sqrt(2)*sigma)
    arg_min = (min - mu) / (np.sqrt(2)*sigma)

    N_window = A * sigma * np.sqrt(np.pi/2) * (erf(arg_max) - erf(arg_min))

    fraction = N_window / N_total

    print("Gaussian counts in window:", N_window)
    print("Fraction kept:", fraction)
counts(popt,dep_min,dep_max)
counts(popt2,fep_min,fep_max)
```
